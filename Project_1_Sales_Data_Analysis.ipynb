```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('sales_data.xlsx',encoding='ISO 8859-1')
```

**DATA CLEANING**

```python
#Removing unwanted columns
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
#Droping Nan values
df.dropna(inplace=True)
```

```python
#Dropping duplicates
df.drop_duplicates(inplace=True)
```

```python
#Changing the values in the Gender column
df['Gender']=df['Gender'].map({'F':'Female','M':'Male'})
```

```python
#Renaming the Marital_Status to Martial
df.rename(columns={'Marital_Status':'Martial'},inplace=True)
```

```python
#Changing the values in Martial column
df['Martial']=df['Martial'].map({0:'Single',1:'Married'})
```

```python
#Resetting the index
df.reset_index(inplace=True)
```

```python
#Dropping the index column
df.drop('index',axis=1,inplace=True)
```

**DATA VISUALIZATION**

```python
a=sns.countplot(x='Gender',data=df,hue='Gender')
def data_label(a):
  for i in a.containers:
    a.bar_label(i)
plt.title('Count by Gender',color='Green')
data_label(a)
```

```python
gen=df.groupby('Gender')['Amount'].sum()
gen=pd.DataFrame(gen)
a=sns.barplot(x='Gender',y='Amount',data=gen,hue='Gender')
plt.title('Amount by Gender',color='Green')
data_label(a)
```

**By analayzing we got to know that Female's are spending more than Male's**

```python
a=sns.countplot(x='Age Group',data=df,hue='Age Group')
plt.title('Count by Age Group',color='Green')
data_label(a)
```

```python
a=df.groupby('Age Group')['Amount'].sum()
age=pd.DataFrame(a)
plt.figure(figsize=(10,5))
a=sns.barplot(x='Age Group',y='Amount',data=age,hue='Age Group')
plt.title('Age group',color='Green')
data_label(a)
```

**By analayzing Age Group column get to know that 26-35 year group people will purchase more**

```python
a=sns.countplot(x='Martial',data=df,hue='Martial')
data_label(a)
```

```python
a=df.groupby('Martial')['Amount'].sum()
martial=pd.DataFrame(a)
martial=sns.barplot(x='Martial',y='Amount',hue='Martial',data=martial)
plt.title('Amount by Martial',color='Green')
data_label(martial)
```

**By analazing that single people will purchase more compared to Married people**

```python
plt.figure(figsize=(12,7))
a=sns.countplot(x='State',data=df,hue='State')
plt.xticks(rotation=90)
plt.title('Count by State',color='Green')
data_label(a)
```

```python
plt.figure(figsize=(15,7))
plt.xticks(rotation=90)
plt.title('Amount by State',color='Green')
a=df.groupby('State')['Amount'].sum()
state=pd.DataFrame(a)
state=sns.barplot(x='State',y='Amount',data=state,hue='State')
data_label(state)
```

**We got to know that Uttar pradesh takes No.1 spot in purchasing**

```python
plt.figure(figsize=(14,9))
plt.xticks(rotation=90)
a=sns.countplot(x='Occupation',data=df,hue='Occupation')
plt.title('Count by occupation',color='Green')
data_label(a)
```

```python
plt.figure(figsize=(15,7))
plt.xticks(rotation=90)
plt.title('Amount by Occupation',color='Green')
a=df.groupby('Occupation')['Amount'].sum()
occ=pd.DataFrame(a)
occ=sns.barplot(x='Occupation',y='Amount',data=occ,hue='Occupation')
data_label(occ)
```

**IT sector people spends most money on purchasing**

```python
plt.figure(figsize=(13,7))
plt.xticks(rotation=90)
plt.title('Count by Product_Category',color='Green')
a=sns.countplot(x='Product_Category',data=df,hue='Product_Category')
data_label(a)
```

```python
plt.figure(figsize=(16,7))
plt.xticks(rotation=90)
plt.title('Amount Product_Category',color='Green')
a=df.groupby('Product_Category')['Amount'].sum()
pro=pd.DataFrame(a)
pro=sns.barplot(x='Product_Category',y='Amount',hue='Product_Category',data=pro)
data_label(pro)
```
